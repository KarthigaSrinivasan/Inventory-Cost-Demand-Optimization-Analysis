```python
import os

folder = "cleaned_data"

os.makedirs(folder, exist_ok=True)

dataframes = {
    "suppliers": df,
    "stores": df1,
    "sales": df2,
    "products": df3,
    "production":df4,
    "inventory": df5
}

for name, df in dataframes.items():
    df.to_csv(
        os.path.join(folder, f"{name}_cleaned.csv"),
        index=False
    )

print("All cleaned CSV files saved successfully!")
```
